```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation


# =========================
# CONFIG
# =========================

OUTPUT_DIR = Path("animations/progress_bars")
OUTPUT_DIR.mkdir(parents=True, exist_ok=True)

OUTPUT_FILE = OUTPUT_DIR / "vertical_glow_progress_bar.gif"

FPS = 30
DURATION_SEC = 6
N_FRAMES = FPS * DURATION_SEC

FIGSIZE = (4, 8)
DPI = 140

BAR_WIDTH = 1.0
BAR_HEIGHT = 6.0

N_BLOCKS = 18
BLOCK_GAP = 0.08
BLOCK_ROUNDING = 0.04

BACKGROUND_COLOR = "#02060A"

FRAME_COLOR = "#1BFFF2"
FRAME_ALPHA = 0.55

EMPTY_COLOR = "#052025"
EMPTY_ALPHA = 0.45

ACTIVE_COLOR = "#26FFF2"
ACTIVE_EDGE_COLOR = "#B8FFFB"

GLOW_COLOR = "#26FFF2"
TEXT_COLOR = "#B8FFFB"

SHOW_PERCENT_TEXT = True


# =========================
# GEOMETRY
# =========================

block_h = (BAR_HEIGHT - BLOCK_GAP * (N_BLOCKS - 1)) / N_BLOCKS
x0 = -BAR_WIDTH / 2
y0 = -BAR_HEIGHT / 2


# =========================
# FIGURE
# =========================

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)
fig.patch.set_facecolor(BACKGROUND_COLOR)
ax.set_facecolor(BACKGROUND_COLOR)

ax.set_xlim(-1.6, 1.6)
ax.set_ylim(-3.6, 3.6)
ax.set_aspect("equal")
ax.axis("off")


# =========================
# STATIC FRAME
# =========================

frame = plt.Rectangle(
    (x0 - 0.12, y0 - 0.12),
    BAR_WIDTH + 0.24,
    BAR_HEIGHT + 0.24,
    fill=False,
    linewidth=2.0,
    edgecolor=FRAME_COLOR,
    alpha=FRAME_ALPHA,
)

ax.add_patch(frame)

# corner accents
corner_len = 0.32
for sx in [-1, 1]:
    for sy in [-1, 1]:
        cx = sx * (BAR_WIDTH / 2 + 0.22)
        cy = sy * (BAR_HEIGHT / 2 + 0.22)

        ax.plot(
            [cx - sx * corner_len, cx],
            [cy, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )
        ax.plot(
            [cx, cx],
            [cy - sy * corner_len, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )


# =========================
# BLOCKS
# =========================

blocks = []
glow_blocks = []

for i in range(N_BLOCKS):
    y = y0 + i * (block_h + BLOCK_GAP)

    glow = plt.Rectangle(
        (x0 - 0.04, y - 0.025),
        BAR_WIDTH + 0.08,
        block_h + 0.05,
        facecolor=GLOW_COLOR,
        edgecolor="none",
        alpha=0.0,
    )
    ax.add_patch(glow)
    glow_blocks.append(glow)

    block = plt.Rectangle(
        (x0, y),
        BAR_WIDTH,
        block_h,
        facecolor=EMPTY_COLOR,
        edgecolor=FRAME_COLOR,
        linewidth=0.8,
        alpha=EMPTY_ALPHA,
    )
    ax.add_patch(block)
    blocks.append(block)


percent_text = ax.text(
    0,
    y0 - 0.55,
    "000%",
    color=TEXT_COLOR,
    fontsize=16,
    ha="center",
    va="center",
    family="monospace",
    alpha=0.9 if SHOW_PERCENT_TEXT else 0.0,
)


# =========================
# ANIMATION
# =========================

def progress_curve(t: float) -> float:
    """
    Smooth 0 -> 1 -> 0 loop.
    t is normalized to [0, 1).
    """
    return 0.5 - 0.5 * np.cos(2 * np.pi * t)


def update(frame_idx):
    t = frame_idx / N_FRAMES
    progress = progress_curve(t)

    active_blocks = int(np.floor(progress * N_BLOCKS))
    pulse = 0.5 + 0.5 * np.sin(2 * np.pi * t * 6)

    for i, block in enumerate(blocks):
        is_active = i < active_blocks
        is_head = i == active_blocks and active_blocks < N_BLOCKS

        if is_active:
            block.set_facecolor(ACTIVE_COLOR)
            block.set_edgecolor(ACTIVE_EDGE_COLOR)
            block.set_alpha(0.75 + 0.25 * pulse)
            glow_blocks[i].set_alpha(0.12 + 0.20 * pulse)
        elif is_head:
            block.set_facecolor(ACTIVE_COLOR)
            block.set_edgecolor(ACTIVE_EDGE_COLOR)
            block.set_alpha(0.35 + 0.35 * pulse)
            glow_blocks[i].set_alpha(0.10 + 0.25 * pulse)
        else:
            block.set_facecolor(EMPTY_COLOR)
            block.set_edgecolor(FRAME_COLOR)
            block.set_alpha(EMPTY_ALPHA)
            glow_blocks[i].set_alpha(0.0)

    frame.set_alpha(FRAME_ALPHA + 0.15 * pulse)

    percent_text.set_text(f"{int(progress * 100):03d}%")

    return [frame, percent_text, *blocks, *glow_blocks]


anim = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    interval=1000 / FPS,
    blit=True,
)

anim.save(
    OUTPUT_FILE,
    writer="pillow",
    fps=FPS,
)

plt.close(fig)

print(f"Saved: {OUTPUT_FILE}")
```

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation, PillowWriter, FFMpegWriter
from matplotlib.patches import Polygon, Rectangle


# =========================
# EXPORT CONFIG
# =========================

OUTPUT_FORMAT = "webm"   # "gif", "webm", "mp4"

OUT_DIR = Path("animations/progress_bars")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_FILE = OUT_DIR / f"vertical_skew_glow_progress_bar.{OUTPUT_FORMAT}"

FPS = 30
DURATION_SEC = 6
N_FRAMES = FPS * DURATION_SEC


# =========================
# FIGURE CONFIG
# =========================

FIGSIZE = (4, 8)
DPI = 140

BACKGROUND_COLOR = "#02060A"


# =========================
# BAR CONFIG
# =========================

BAR_WIDTH = 1.0
BAR_HEIGHT = 6.0

N_BLOCKS = 18
BLOCK_GAP = 0.08
BLOCK_SKEW = 0.22

FRAME_PADDING_X = 0.16
FRAME_PADDING_Y = 0.16

FRAME_COLOR = "#1BFFF2"
FRAME_ALPHA = 0.55

EMPTY_COLOR = "#052025"
EMPTY_ALPHA = 0.45

ACTIVE_COLOR = "#26FFF2"
ACTIVE_EDGE_COLOR = "#B8FFFB"

GLOW_COLOR = "#26FFF2"
TEXT_COLOR = "#B8FFFB"

SHOW_PERCENT_TEXT = True


# =========================
# GEOMETRY
# =========================

usable_width = BAR_WIDTH - BLOCK_SKEW
block_h = (BAR_HEIGHT - BLOCK_GAP * (N_BLOCKS - 1)) / N_BLOCKS

x0 = -BAR_WIDTH / 2
y0 = -BAR_HEIGHT / 2

frame_x = x0 - FRAME_PADDING_X
frame_y = y0 - FRAME_PADDING_Y
frame_w = BAR_WIDTH + FRAME_PADDING_X * 2
frame_h = BAR_HEIGHT + FRAME_PADDING_Y * 2


# =========================
# FIGURE
# =========================

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)
fig.patch.set_facecolor(BACKGROUND_COLOR)
ax.set_facecolor(BACKGROUND_COLOR)

ax.set_xlim(-1.6, 1.6)
ax.set_ylim(-3.8, 3.8)
ax.set_aspect("equal")
ax.axis("off")


# =========================
# STATIC FRAME
# =========================

frame = Rectangle(
    (frame_x, frame_y),
    frame_w,
    frame_h,
    fill=False,
    linewidth=2.0,
    edgecolor=FRAME_COLOR,
    alpha=FRAME_ALPHA,
)

ax.add_patch(frame)

corner_len = 0.32

for sx in [-1, 1]:
    for sy in [-1, 1]:
        cx = sx * (frame_w / 2 + FRAME_PADDING_X)
        cy = sy * (frame_h / 2 + FRAME_PADDING_Y)

        ax.plot(
            [cx - sx * corner_len, cx],
            [cy, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )
        ax.plot(
            [cx, cx],
            [cy - sy * corner_len, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )


# =========================
# BLOCKS
# =========================

blocks = []
glow_blocks = []

for i in range(N_BLOCKS):
    y = y0 + i * (block_h + BLOCK_GAP)

    glow_pts = [
        [x0 - 0.03, y - 0.02],
        [x0 + usable_width + 0.03, y - 0.02],
        [x0 + usable_width + BLOCK_SKEW + 0.03, y + block_h + 0.02],
        [x0 + BLOCK_SKEW - 0.03, y + block_h + 0.02],
    ]

    glow = Polygon(
        glow_pts,
        closed=True,
        facecolor=GLOW_COLOR,
        edgecolor="none",
        alpha=0.0,
    )

    ax.add_patch(glow)
    glow_blocks.append(glow)

    block_pts = [
        [x0, y],
        [x0 + usable_width, y],
        [x0 + usable_width + BLOCK_SKEW, y + block_h],
        [x0 + BLOCK_SKEW, y + block_h],
    ]

    block = Polygon(
        block_pts,
        closed=True,
        facecolor=EMPTY_COLOR,
        edgecolor=FRAME_COLOR,
        linewidth=0.8,
        alpha=EMPTY_ALPHA,
        joinstyle="miter",
    )

    ax.add_patch(block)
    blocks.append(block)


percent_text = ax.text(
    0,
    frame_y - 0.42,
    "000%",
    color=TEXT_COLOR,
    fontsize=16,
    ha="center",
    va="center",
    family="monospace",
    alpha=0.9 if SHOW_PERCENT_TEXT else 0.0,
)


# =========================
# ANIMATION
# =========================

def progress_curve(t: float) -> float:
    return min(1.0, max(0.0, t))


def update(frame_idx):
    t = frame_idx / max(1, N_FRAMES - 1)
    progress = progress_curve(t)

    active_blocks = int(np.floor(progress * N_BLOCKS))
    pulse = 0.5 + 0.5 * np.sin(2 * np.pi * t * 6)

    for i, block in enumerate(blocks):
        is_active = i < active_blocks
        is_head = i == active_blocks and active_blocks < N_BLOCKS

        if is_active:
            block.set_facecolor(ACTIVE_COLOR)
            block.set_edgecolor(ACTIVE_EDGE_COLOR)
            block.set_alpha(0.75 + 0.25 * pulse)
            glow_blocks[i].set_alpha(0.12 + 0.20 * pulse)

        elif is_head:
            block.set_facecolor(ACTIVE_COLOR)
            block.set_edgecolor(ACTIVE_EDGE_COLOR)
            block.set_alpha(0.35 + 0.35 * pulse)
            glow_blocks[i].set_alpha(0.10 + 0.25 * pulse)

        else:
            block.set_facecolor(EMPTY_COLOR)
            block.set_edgecolor(FRAME_COLOR)
            block.set_alpha(EMPTY_ALPHA)
            glow_blocks[i].set_alpha(0.0)

    frame.set_alpha(FRAME_ALPHA + 0.15 * pulse)
    percent_text.set_text(f"{int(progress * 100):03d}%")

    return [frame, percent_text, *blocks, *glow_blocks]


anim = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    interval=1000 / FPS,
    blit=True,
)


# =========================
# EXPORT
# =========================

if OUTPUT_FORMAT == "gif":
    writer = PillowWriter(fps=FPS)
    anim.save(OUT_FILE, writer=writer)

elif OUTPUT_FORMAT == "webm":
    writer = FFMpegWriter(
        fps=FPS,
        codec="libvpx-vp9",
        bitrate=1800,
        extra_args=[
            "-pix_fmt", "yuv420p",
            "-auto-alt-ref", "0",
        ],
    )
    anim.save(OUT_FILE, writer=writer, dpi=DPI)

elif OUTPUT_FORMAT == "mp4":
    writer = FFMpegWriter(
        fps=FPS,
        codec="libx264",
        bitrate=1800,
        extra_args=[
            "-pix_fmt", "yuv420p",
            "-movflags", "+faststart",
        ],
    )
    anim.save(OUT_FILE, writer=writer, dpi=DPI)

else:
    raise ValueError(f"Unsupported OUTPUT_FORMAT: {OUTPUT_FORMAT}")

plt.close(fig)

print(f"Saved: {OUT_FILE}")
```

```python
from pathlib import Path

import numpy as np
import matplotlib.pyplot as plt

from matplotlib.animation import FuncAnimation, PillowWriter, FFMpegWriter
from matplotlib.patches import Circle, Rectangle


# =========================
# EXPORT CONFIG
# =========================

OUTPUT_FORMAT = "webm"   # "gif", "webm", "mp4"

OUT_DIR = Path("animations/progress_bars")
OUT_DIR.mkdir(parents=True, exist_ok=True)

OUT_FILE = OUT_DIR / f"vertical_dot_progress_bar.{OUTPUT_FORMAT}"

FPS = 30
DURATION_SEC = 6
N_FRAMES = FPS * DURATION_SEC


# =========================
# FIGURE CONFIG
# =========================

FIGSIZE = (4, 8)
DPI = 140

BACKGROUND_COLOR = "#02060A"


# =========================
# DOT BAR CONFIG
# =========================

BAR_WIDTH = 1.0
BAR_HEIGHT = 6.8

N_DOTS = 22
DOT_DIAMETER = 0.20
DOT_GAP = 0.08

FRAME_PADDING_X = 0.22
FRAME_PADDING_Y = 0.18

FRAME_COLOR = "#1BFFF2"
FRAME_ALPHA = 0.55

EMPTY_DOT_COLOR = "#052025"
EMPTY_DOT_ALPHA = 0.45

ACTIVE_DOT_COLOR = "#26FFF2"
ACTIVE_EDGE_COLOR = "#B8FFFB"

GLOW_COLOR = "#26FFF2"
GLOW_DIAMETER_MULT = 2.2

TEXT_COLOR = "#B8FFFB"

SHOW_PERCENT_TEXT = True


# =========================
# GEOMETRY
# =========================

DOT_RADIUS = DOT_DIAMETER / 2
GLOW_RADIUS = DOT_RADIUS * GLOW_DIAMETER_MULT

usable_height = BAR_HEIGHT
total_dots_height = N_DOTS * DOT_DIAMETER + (N_DOTS - 1) * DOT_GAP

if total_dots_height > usable_height:
    raise ValueError(
        f"Точки не помещаются в BAR_HEIGHT. "
        f"Нужно {total_dots_height:.2f}, доступно {usable_height:.2f}. "
        f"Уменьши N_DOTS, DOT_DIAMETER или DOT_GAP."
    )

x_center = 0.0
y0 = -BAR_HEIGHT / 2
first_dot_y = y0 + DOT_RADIUS + (usable_height - total_dots_height) / 2

frame_x = -BAR_WIDTH / 2 - FRAME_PADDING_X
frame_y = -BAR_HEIGHT / 2 - FRAME_PADDING_Y
frame_w = BAR_WIDTH + FRAME_PADDING_X * 2
frame_h = BAR_HEIGHT + FRAME_PADDING_Y * 2


# =========================
# FIGURE
# =========================

fig, ax = plt.subplots(figsize=FIGSIZE, dpi=DPI)
fig.patch.set_facecolor(BACKGROUND_COLOR)
ax.set_facecolor(BACKGROUND_COLOR)

ax.set_xlim(-1.6, 1.6)
ax.set_ylim(-3.8, 3.8)
ax.set_aspect("equal")
ax.axis("off")


# =========================
# STATIC FRAME
# =========================

frame = Rectangle(
    (frame_x, frame_y),
    frame_w,
    frame_h,
    fill=False,
    linewidth=2.0,
    edgecolor=FRAME_COLOR,
    alpha=FRAME_ALPHA,
)

ax.add_patch(frame)

corner_len = 0.32

for sx in [-1, 1]:
    for sy in [-1, 1]:
        cx = sx * (frame_w / 2 + FRAME_PADDING_X)
        cy = sy * (frame_h / 2 + FRAME_PADDING_Y)

        ax.plot(
            [cx - sx * corner_len, cx],
            [cy, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )
        ax.plot(
            [cx, cx],
            [cy - sy * corner_len, cy],
            color=FRAME_COLOR,
            linewidth=2.0,
            alpha=0.9,
        )


# =========================
# DOTS
# =========================

dots = []
glow_dots = []

for i in range(N_DOTS):
    y = first_dot_y + i * (DOT_DIAMETER + DOT_GAP)

    glow = Circle(
        (x_center, y),
        radius=GLOW_RADIUS,
        facecolor=GLOW_COLOR,
        edgecolor="none",
        alpha=0.0,
    )
    ax.add_patch(glow)
    glow_dots.append(glow)

    dot = Circle(
        (x_center, y),
        radius=DOT_RADIUS,
        facecolor=EMPTY_DOT_COLOR,
        edgecolor=FRAME_COLOR,
        linewidth=0.8,
        alpha=EMPTY_DOT_ALPHA,
    )
    ax.add_patch(dot)
    dots.append(dot)


percent_text = ax.text(
    0,
    frame_y - 0.42,
    "000%",
    color=TEXT_COLOR,
    fontsize=16,
    ha="center",
    va="center",
    family="monospace",
    alpha=0.9 if SHOW_PERCENT_TEXT else 0.0,
)


# =========================
# ANIMATION
# =========================

def progress_curve(t: float) -> float:
    return min(1.0, max(0.0, t))


def update(frame_idx):
    t = frame_idx / max(1, N_FRAMES - 1)
    progress = progress_curve(t)

    active_dots = int(np.floor(progress * N_DOTS))
    pulse = 0.5 + 0.5 * np.sin(2 * np.pi * t * 6)

    for i, dot in enumerate(dots):
        is_active = i < active_dots
        is_head = i == active_dots and active_dots < N_DOTS

        if is_active:
            dot.set_facecolor(ACTIVE_DOT_COLOR)
            dot.set_edgecolor(ACTIVE_EDGE_COLOR)
            dot.set_alpha(0.75 + 0.25 * pulse)
            glow_dots[i].set_alpha(0.10 + 0.22 * pulse)

        elif is_head:
            dot.set_facecolor(ACTIVE_DOT_COLOR)
            dot.set_edgecolor(ACTIVE_EDGE_COLOR)
            dot.set_alpha(0.35 + 0.35 * pulse)
            glow_dots[i].set_alpha(0.08 + 0.25 * pulse)

        else:
            dot.set_facecolor(EMPTY_DOT_COLOR)
            dot.set_edgecolor(FRAME_COLOR)
            dot.set_alpha(EMPTY_DOT_ALPHA)
            glow_dots[i].set_alpha(0.0)

    frame.set_alpha(FRAME_ALPHA + 0.15 * pulse)
    percent_text.set_text(f"{int(progress * 100):03d}%")

    return [frame, percent_text, *dots, *glow_dots]


anim = FuncAnimation(
    fig,
    update,
    frames=N_FRAMES,
    interval=1000 / FPS,
    blit=True,
)


# =========================
# EXPORT
# =========================

if OUTPUT_FORMAT == "gif":
    writer = PillowWriter(fps=FPS)
    anim.save(OUT_FILE, writer=writer)

elif OUTPUT_FORMAT == "webm":
    writer = FFMpegWriter(
        fps=FPS,
        codec="libvpx-vp9",
        bitrate=1800,
        extra_args=[
            "-pix_fmt", "yuv420p",
            "-auto-alt-ref", "0",
        ],
    )
    anim.save(OUT_FILE, writer=writer, dpi=DPI)

elif OUTPUT_FORMAT == "mp4":
    writer = FFMpegWriter(
        fps=FPS,
        codec="libx264",
        bitrate=1800,
        extra_args=[
            "-pix_fmt", "yuv420p",
            "-movflags", "+faststart",
        ],
    )
    anim.save(OUT_FILE, writer=writer, dpi=DPI)

else:
    raise ValueError(f"Unsupported OUTPUT_FORMAT: {OUTPUT_FORMAT}")

plt.close(fig)

print(f"Saved: {OUT_FILE}")
```
